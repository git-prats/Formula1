# Set-up the project environment for F1 Project
1. Create External Location f1-ext-dl-formula1
2. Create Catalog formula1
3. Create Schemas landing, bronze, silver and gold
4. Create Volume Files in the landing schema

### Create External Location

```sql
CREATE EXTERNAL LOCATION IF NOT EXISTS f1_dl_formula1
URL 'abfss://formula1@f1extdl.dfs.core.windows.net/'
WITH (STORAGE CREDENTIAL `f1-sc`)
COMMENT 'External location for the formula1 container';
```

### Access Cloud Storage

```sql
%fs ls 'abfss://formula1@f1extdl.dfs.core.windows.net/landing'
```

### Create Catalog F1

```sql
SHOW CATALOGS;
```

```sql
CREATE CATALOG IF NOT EXISTS f1
   MANAGED LOCATION 'abfss://formula1@f1extdl.dfs.core.windows.net/' 
   COMMENT 'This is the main catalog for the F1 project' ;
```

### Create Schemas landing, bronze, silver, gold

```sql
CREATE SCHEMA IF NOT EXISTS f1.landing;
CREATE SCHEMA IF NOT EXISTS f1.bronze
    MANAGED LOCATION 'abfss://formula1@f1extdl.dfs.core.windows.net/bronze';
CREATE SCHEMA IF NOT EXISTS f1.silver
    MANAGED LOCATION 'abfss://formula1@f1extdl.dfs.core.windows.net/silver';
CREATE SCHEMA IF NOT EXISTS f1.gold
    MANAGED LOCATION 'abfss://formula1@f1extdl.dfs.core.windows.net/gold'; 
```

```sql
SELECT current_catalog();
```

```sql
USE CATALOG f1;
```

```sql
SHOW SCHEMAS;
```

### Create Volume Files


```sql
CREATE EXTERNAL VOLUME IF NOT EXISTS f1.landing.files
LOCATION 'abfss://formula1@f1extdl.dfs.core.windows.net/landing';
```
